```python
#nested objects behaviour and shallow copy and deepcopy

xv = [3,4,5,6,[99]]
cd = xv
print(id(cd[-1]) ,  id(xv[-1]))  # the total object (inner , outer object are same) becoz we use refrence '=' to copy that 
print(id(cd),id(xv))

#but in sahllow copy and deep coipy you see diff

import copy as cp

list1 = [[1],3,5,10]
list2 = cp.copy(list1)

print('Before change')
print(list2 , "duplicated") #[[1], 3, 5, 10] duplicated
print(list1 , "original") #[[1], 3, 5, 10] original

list2[0].append(8)   # as you see the copy changes the inner object because it is not copied it is pointing to same variable ( i mean
# referenced only)


print('After change')
print(list2 , "duplicated") #[[1, 8], 3, 5, 10] duplicated
print(list1 , "original") #[[1, 8], 3, 5, 10] original

#try with deep copy now

list3 = cp.deepcopy(list2)

print('Deepcopy')

print(list3 , "duplicated") #[[1, 8], 3, 5, 10] duplicated
print(list2 , "original")

#change -->
list3[0].append(0)

print(list3 , "duplicated") #[[1, 8, 0], 3, 5, 10] duplicated
print(list2 , "original") # [[1, 8], 3, 5, 10] original

#☝🏻 why original's nested doesnt change this time becaz this is deepcopy --> this is deeply copied and newed totally


```
